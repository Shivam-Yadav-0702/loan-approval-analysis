
# Loan Approval Analysis

### Exploratory Data Analysis Project

This project focuses on analyzing a loan approval dataset to understand:
- customer demographics
- income patterns
- loan approval trends
- relationships between variables

The main goal is to perform Exploratory Data Analysis (EDA) and create visualizations that help understand the data better.

---

## Objectives
- Perform data cleaning
- Handle missing values
- Analyze numeric and categorical variables
- Create visualizations
- Find useful business insights


```

# importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# display settings
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10,5)

print("Libraries imported successfully")
```

```

# loading dataset

df = pd.read_csv(r"/mnt/data/loan_sanction_test.csv")

print("Dataset Loaded Successfully")

# checking first 5 rows
df.head()
```


## Dataset Overview

First, I explored the dataset structure to understand:
- number of rows and columns
- data types
- missing values
- overall quality of data


```

# dataset information

print("Shape of Dataset:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```

# handling missing values

# filling categorical missing values with mode
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

# filling numeric missing values with median
for col in df.select_dtypes(include=['int64','float64']).columns:
    df[col].fillna(df[col].median(), inplace=True)

print("Missing values handled successfully")
```


## Statistical Summary

This section helps in understanding:
- average values
- data spread
- minimum and maximum values
- possible outliers


```

# descriptive statistics

df.describe()
```


# Univariate Analysis

In this section, individual variables are analyzed separately using:
- histograms
- boxplots
- countplots


```

# histogram for applicant income

plt.figure(figsize=(10,5))
sns.histplot(df['ApplicantIncome'], bins=30)

plt.title("Distribution of Applicant Income")
plt.xlabel("Applicant Income")
plt.ylabel("Frequency")
plt.show()

print("Insight:")
print("Most applicants fall in the lower income range with few high-income outliers.")
```

```

# boxplot for loan amount

plt.figure(figsize=(8,5))
sns.boxplot(x=df['LoanAmount'])

plt.title("Loan Amount Boxplot")
plt.show()

print("Observation:")
print("There are several outliers in loan amount values.")
```

```

# loan approval status count

plt.figure(figsize=(6,4))
sns.countplot(x='Loan_Status', data=df)

plt.title("Loan Approval Distribution")
plt.show()

print("Insight:")
print("Approved loans are higher compared to rejected applications.")
```


# Bivariate Analysis

This section explores relationships between variables.


```

# applicant income vs loan amount

plt.figure(figsize=(8,5))
sns.scatterplot(x='ApplicantIncome', y='LoanAmount', data=df)

plt.title("Applicant Income vs Loan Amount")
plt.show()

print("Observation:")
print("Higher income applicants generally apply for larger loan amounts.")
```

```

# education vs loan amount

plt.figure(figsize=(7,5))
sns.boxplot(x='Education', y='LoanAmount', data=df)

plt.title("Education vs Loan Amount")
plt.show()

print("Insight:")
print("Graduate applicants tend to apply for slightly higher loan amounts.")
```


# Multivariate Analysis

Correlation analysis helps understand relationships between numeric variables.


```

# correlation heatmap

numeric_df = df.select_dtypes(include=['int64','float64'])

correlation = numeric_df.corr()

plt.figure(figsize=(10,6))
sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()

print("Observation:")
print("Applicant income and loan amount show moderate positive correlation.")
```

```

# pairplot

sns.pairplot(numeric_df)
plt.show()
```


# Final Conclusion

### Key Findings
- Most loan applicants belong to lower and middle income groups.
- Loan approvals are more common than rejections.
- Applicant income affects loan amount significantly.
- Some columns contain outliers which may affect future modeling.

### Recommendations
- Financial institutions should improve risk analysis for outlier applicants.
- More detailed credit history analysis can improve approval accuracy.
- Additional customer demographic analysis may provide better business insights.
